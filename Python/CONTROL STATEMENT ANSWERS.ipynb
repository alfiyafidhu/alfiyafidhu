**PART A : WHILE LOOP**

```python
#1
i=1
while i<=7:
    print("Day",i)
    i+=1
```

```python
#2
sales = [1200, 1500, 900, 2000, 1800]

i = 0

while i < len(sales):
    print(sales[i])
    i += 1     
```

```python
#3
sales = [1200, 1500, 900, 2000, 1800]

i = 0
total = 0

while i < len(sales):
    total += sales[i]
    i += 1

print("Total sales:", total)
```

```python
#4
marks = [45, 78, 62, 90, 55]

i = 0
total = 0

while i < len(marks):
    total = total + marks[i]
    i += 1

average = total / len(marks)

print("Average marks:", average)
```

```python
#5
marks = [45, 78, 62, 90, 55]

i = 0
count = 0

while i < len(marks):
    if marks[i] >= 50:
        count += 1
    i += 1

print("Number of passed students:", count)
```

```python
#6
quantity = int(input("Enter quantity: "))

i = 1

while i <= quantity:
    print(25 * i)
    i += 1
```

```python
#7
temperatures = [31, 34, 29, 36, 33]

i = 1
highest = temperatures[0]

while i < len(temperatures):
    if temperatures[i] > highest:
        highest = temperatures[i]
    i += 1

print("Highest temperature:", highest)
```

**PART B: WHILE WITH ELSE**

```python
#8
month = 1

while month <= 3:
    print(month)
    month += 1
else:
    print("Report Generated")
```

```python
#9
ids = [101, 102, 103, 104]

i = 0

while i < len(ids):
    print(ids[i])
    i += 1
else:
    print("All customers listed")
```
