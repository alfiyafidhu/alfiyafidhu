**Part A: while + break (stop when you find what you need)**

```python
#1 The daily sales of a shop for 5 days are given. 
#Find the first day on which the sales were more than 2000. Print the day number and the sales, then stop checking.

sales = [1200, 1500, 900, 2500, 1800]

i=0
n=len(sales)
print(n)
while i<=n:
     if sales[i]>=2000:
      print("Highest sales day: ",i+1)
      print("day number: ",sales[i])
      break
     i+=1
```

```python
#2The marks of 5 students are given. Find the first student who failed (marks below 35). 
#Print the student number and the marks, then stop.
marks = [78, 65, 90, 28, 55]
i=0
n=len(marks)

while i<=n:
    if marks[i]<=35:
        print("Failed Student number: ",i+1)
        print("Marks: ",marks[i])
        break
    i+=1
```

```python
#3.Add the daily sales one by one. Stop as soon as the running total becomes 5000 or more. 
#Print the day number and the total at that point.

sales = [1200, 1500, 900, 2000, 1800]

i = 0
total = 0

while i < len(sales):
    total += sales[i]

    if total >= 5000:
        print("Day:", i + 1)
        print("Total:", total)
        break

    i += 1
    
    
    
```

**Part C: while + break + continue together**


```python
#7.A data list has missing values (None), and the value -1 marks the end of the data. 
#Skip the None values, stop when -1 is found, and print the total of the valid values before it.

data = [100, None, 250, None, 300, -1, 400]

i=0
total=0

while i<len(data):
    if data[i]==None:
        i+=1
        continue
        
    if data[i]==-1:
     break
    total += data[i]
    i+=1

print("Total of valid values = ",total)
```

```python
#8.The shop was closed on some days, and the sales for those days are stored as 0. 
#Skip the days with 0 sales and print the other sales. Stop the loop when a sale of more than 2000 is found,
#and print "Target crossed".
sales = [1200, 0, 1500, 0, 2500, 1800]
i = 0
n = len(sales)

while i < n:
    if sales[i] == 0:
        i += 1
        continue

    if sales[i] > 2000:
        print("Target crossed")
        break
          i += 1

    print("Sale:", sales[i])
  
```

9.Write a program that keeps asking the user to enter a sale amount and adds it to the total:
•  If the user enters a negative number, print "Invalid amount" and skip it. (Use continue.)
•  If the user enters 0, stop the program. (Use break.)
•  Otherwise, add the amount to the total.
At the end, print the total sales. (Hint: use while True:)


```python

total = 0

while True:  #keeps asking the user for sale amount until the loop is stopped.
    sales = float(input("Enter the sales amount(0 to stop): "))
    if sales<0:
        print("Invalid Amount")
        
        continue

    total+=sales
    if sales==0:
        break

print(total,"is the Total sales")
```

**Part D: while with else + break**


10.A list of customer IDs is given. Take an ID from the user and search for it using a while loop.
ids = [101, 102, 103, 104]
•  If the ID is found, print its position and stop. (Use break.)
•  If the ID is not found, print "Customer not found" using the else block.


```python
ids = [101,102,103,104]
i=0
user = int(input("Enter User ID: "))

while i<len(ids):
    if user==ids[i]:
        print("ID position: ",i+1)  #i+1 displays position
        break
    i+=1
else:
        print("Customer not found")
```
