```python
#D3
#MONTHLY SALES REPORT DATA
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 9000, 18000, 21000, 17000]
```

```python
total = sum(sales)
print("Total Sales: ",sales)
average = total/len(sales)
print("Average Sales: ",round(average,2))
```

```python
highest_sales = max(sales)
highest_index = sales.index(highest_sales)
highest_month = months[highest_index]
print("Highest Sales happened in", highest_sales,"during",highest_month)
```

```python
#D6
#EVEN AND ODD SEPARATOR
numbers = [12, 7, 8, 3, 20, 15, 6, 9]

even = []
odd = []

for number in numbers:
    if number % 2 == 0:
        even.append(number)
    else:
        odd.append(number)
even.sort()
odd.sort()

print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
#D7
#PALINDROME LIST
for a in ([1, 2, 3, 2, 1], ["a","b","c"]):
    if a == a[::-1]:
        print("Palindrome")
    else:
        print("Not a Palindrome")
```

```python
#TO-DO LIST PROGRAM
tasks = []

while True:
    print("\n1. Add task")
    print("2. Remove task")
    print("3. View tasks")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        task = input("Enter task: ")
        tasks.append(task)
        print("Task added.")

    elif choice == "2":
        task = input("Enter task to remove: ")

        if task in tasks:
            tasks.remove(task)
            print("Task removed.")
        else:
            print("Task not found")

    elif choice == "3":
        print("Tasks:", tasks)

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice")
```
