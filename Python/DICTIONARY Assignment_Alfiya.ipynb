```python
#1 RECORD BUILDER
record ={}
while True:
    key= input("Enter a key(or type stop): ")
    if key == "stop":
        break

    value = input("Enter value: ")


    record.update({key: value})
print("Final Record:",record)
    
```

```python
#2 WORD COUNTER
sentence = "to be or not to be"

words = sentence.split()

word_count = {}

for word in words:
    if word in word_count:
        word_count.update({word: word_count[word] + 1})
    else:
        word_count.update({word: 1})

print(word_count)
```

```python
#3 FAIL FILTER
marks = {"Asha": 35, "Ravi": 78, "Meena": 29, "John": 60}
```

```python
marks = {
    "Asha": 35,
    "Ravi": 78,
    "Meena": 29,
    "John": 60
}

for student in list(marks.keys()):
    if marks[student] < 40:
        marks.pop(student)

print(marks)
```

```python
#5 SHALLOW COPY PUZZLE
a = {"name": "John", "marks": [80, 90]}
b = a.copy()
b["marks"].append(100)
b["name"] = "Sam"
print(a)
# original dictionary changes if a nested list/dictionary is changed in a copy()
# changing a direct value won't change the original dictionary
 
```
