```python
#8-UNIQUE PAIRS USING FROZENSET
pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]
```

a) If I put the tuples directly into a set, I get 5 unique pairs because Python treats ("Asha", "Bala") and ("Bala", "Asha") as different tuples. This is not correct for this problem because a conversation between Asha and Bala should be considered the same regardless of who is listed first.

b) If I convert each pair into a frozenset and store them in an outer set, I get 3 unique conversations. The conversations are between Asha and Bala, Chitra and Dinesh, and Asha and Chitra.

c) I cannot use a normal set inside the outer set because a normal set is mutable and unhashable, so it cannot be an element of another set. A frozenset is immutable and hashable, so it can be stored inside an outer set.
