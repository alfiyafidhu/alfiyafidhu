```python
#5
num = int(input("Enter a three-digit number: "))

hundreds = num // 100
tens = (num // 10) % 10
ones = num % 10

reverse = ones * 100 + tens * 10 + hundreds

if num == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
#6
cost_price = float(input("Enter cost price: "))
selling_price = float(input("Enter selling price: "))

if selling_price > cost_price:
    profit = selling_price - cost_price
    percentage = (profit / cost_price) * 100
    print("Profit:", profit)
    print("Profit percentage:", percentage)

elif selling_price < cost_price:
    loss = cost_price - selling_price
    percentage = (loss / cost_price) * 100
    print("Loss:", loss)
    print("Loss percentage:", percentage)

else:
    print("No profit no loss")
```

```python
#10
distance = float(input("Enter distance in km: "))
hour = int(input("Enter hour (0 to 23): "))

if distance <= 2:
    fare = 60
else:
    extra_km = distance - 2
    fare = 60 + extra_km * 15

if hour >= 22 or hour < 5:
    fare = fare + (fare * 25 / 100)

print("Final fare:", fare)
```

```python
#11
x = float(input("Enter x: "))
y = float(input("Enter y: "))

if x == 0 and y == 0:
    print("At the origin")

elif y == 0:
    print("On the x-axis")

elif x == 0:
    print("On the y-axis")

elif x > 0 and y > 0:
    print("Quadrant 1")

elif x < 0 and y > 0:
    print("Quadrant 2")

elif x < 0 and y < 0:
    print("Quadrant 3")

else:
    print("Quadrant 4")
```
